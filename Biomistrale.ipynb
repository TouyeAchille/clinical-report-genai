```python
from vllm import LLM, SamplingParams

print("Chargement de BioMistral en version compressée AWQ (4-bits)...")

# On utilise la version pré-compressée officielle de BioMistral
# et on indique à vLLM d'utiliser son moteur "awq" natif.
llm = LLM(
    model="BioMistral/BioMistral-7B-AWQ-QGS128-W4-GEMM",
    quantization="awq",
    max_model_len=1024,
    enforce_eager=True
)

print("Modèle chargé avec succès !")

```

```python

parametres = SamplingParams(temperature=0.3, max_tokens=200)
ma_question = ["[INST] Que faire pour être en bonne santé [/INST]"]


resultats = llm.generate(ma_question, parametres)


for output in resultats:
    print("\n--- RÉPONSE DE BIOMISTRAL ---")
    print(output.outputs[0].text)
```
